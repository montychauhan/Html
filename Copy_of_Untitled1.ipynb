```python
from sklearn.datasets import make_classification
import torch
```

```python
#step 1: create a synthetic classification dataset using sklearn
X,y = make_classification(
    n_samples=10, #no of samples
    n_features=2, # no of features
    n_informative=2, # no of informative features
    n_redundant=0, # no of redundant features
    n_classes=2, # no of classes
    random_state=42 #no of reproduciblity
)
```

```python
#convert the data to pytorch tensor
x = torch.tensor(X,dtype=torch.float32)
print(x)
y = torch.tensor(y,dtype=torch.long)
print(y)
```

```python
from torch.utils.data import Dataset,DataLoader
```

```python
class CustomDataset(Dataset):

  def __init__(self,features,labels):
    self.features = features
    self.labels = labels

  def __len__(self):

    return self.features.shape[0]

  def __getitem__(self,index):

    return self.features[index],self.labels[index]
```

```python
dataset = CustomDataset(X,y)
```

```python
len(dataset)
```

```python
dataloader = DataLoader(dataset,batch_size=2,shuffle=True)
```

```python
for batch_features,batch_labels in dataloader:

  print(batch_features)
  print(batch_labels)
  print("-"*50)
```
